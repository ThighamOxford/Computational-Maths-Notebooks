This gives us a neat way to compute Pascal's triangle iteratively.

Code block 7.1 in notes:

```python
N = 10 # number of rows to compute

row = [1] # starting row
rows = [] # this list will contain all rows computed

for n in range(N):
    # Calculate the next row from the previous one
    row = [([0] + row)[i] + (row + [0])[i] for i in range(n+1)]
    rows.append(row)

# Print the output
for row in rows:
    print(row)
```

In this Pascal example we represent polynomials in the monomial basis and using Python lists.

For example $1 + 3x + 8x^3$ would be represented by $[1, 3, 0, 8]$.

Given a representation of a polynomial in this manner, write a **for** loop to generate the representation of its derivative.

[Note to self: mention private and global variables.]

```python

def derivative(polynomial):
    """  
    A function to calculate the derivative of a polynomial
        Input: polynomial -  a list of monomial basis coefficients (e.g [1,2,3])

        Output: derivative - a list in same basis. 
    """

    derivative = []

    for i in range( 1, len(polynomial) ):
        derivative.append( polynomial[i] *i )

    return derivative

p4 = rows[3] 
print("The polynomial given by the fourth row of Pascal's triangle is", p4)
print("The derivative of this polynomial is", derivative(p4))

p7 = rows[6]
print("\n Another example is ",p7,"--->", derivative(p7))
```

#### Exercise 7.4

We now discuss a polynomial representation in a dictionary as opposed to a list - this is better for a polynomial with sparse coefficients such as $ 1 + x^{1000}$.

Given a dictionary with key-value pairs {degree:coefficient}, build a dictionary representation of its derivative.

```python

def derivative_dictionary(polynomial):
    """
    Derivative of a polynomial in dictionary format
        Input - dictionary{degree:coefficient}
        Output - dictionary{degree:coefficient}
    """

    derivative = {degree-1: coefficient*degree for degree, coefficient in polynomial.items() if degree !=0}
    
    return derivative

poly = {0:1, 1:3, 3:8} # 1 + 3x + 8x^3

der = derivative_dictionary(poly)

print(der)
```
