### Exercise 4.1

Compute the last ten digits of $55^{55}$ (as an integer without casting to a string)

```python
""" Printing object types """

print( "1.1 has type", type(1.1) )

print(" 'hello' has type", type("hello"))

print("1 has type", type(1))


# The unsophisticated approach - casting to a string

answer = 55**55
s = str(answer)
print('Using string method, the last ten digits of 55**55 are', s[-10:])
```

### Solution 

We can decompose $55^{55}$ as a sum $A + B = 55^{55}$ where $B$ is the last ten digits of $55^{55}$ and the last ten digits of $A$ are $0$'s.

Two equivalent ways to think of this:

$$ 55^{55}  =  B \quad (\text{mod} \; 10^{10}). $$

or 

$$ 55^{55} = q \times 10^{10} + B $$
where $q$ is the integer part of division and $B$ the remainder.


Python usefully has the command % for computing the remainder on a division - make use of this!

```python
# A better approach

B = answer % 10**10
print('Using modular arithmetic, the last ten digits of 55**55 are', B)
```

### Exercise 4.3

The World Geodetic System is a set of international standards for describing the shape of the Earth. The Earth's geoid is approximated by a reference ellipsoid that takes the form of an oblate shperoid with semi-major and semi-minor axes
$$ a= 6378137.0 \; \text{m} \quad \text{and} \quad c = 6356752.314245 \; \text{m} $$

Use the formula for the surface area of an oblate spheroid,
$$ S = 2 \pi a^2 (1 + \frac{1 - e^2}{e} \text{a}\tanh(e)), \quad \text{where} \; e^2 = 1 - \frac{c^2}{a^2}$$

to calculate the surface area of this reference ellipsoid. Compare it to the surface area of a spherical model of the Earth, with radius $6371$ km.

```python
import numpy as np
import math

# Setting up variables 
a = 6378137.0 # .0 makes this a float
c = 6356752.314245
E2 = 1 - (c**2)/(a**2) 
E = np.sqrt(E2)

# Calculating surface area of oblate spheroid
S = 2 * np.pi * a**2 * (1 +  math.atanh(E)*(1-E2)/E )

### Take out a * and see what error Python calls

# Calculating surface area of spherical model
r = 6371 * 10**3  # convert km to m 
S_sphere = 4 * np.pi * r**2

# Difference in surface area in metres
diff = (S - S_sphere) / 1000 

print('Surface area of oblate spheroid is', f"{S:.2e}", 'm^2')
print('Surface area of spherical model is', f"{S_sphere:.2e}", 'm^2')

print('The oblate spheroid model is', f"{diff:.2e}", "km larger in surface area that the spherical model.")
```

### Exercise 4.5 

If we list all the natural numbers below $10$ that are multiples of $3$ or $5$, we get $3, 5, 6$ and $9$. The sum of these multiples is $23$.

Find the sum of all multiples of $3$ or $5$ below $1000$.


----------------
Since $1000$ is not too large a number we can construct a straight forward algorithm to check whether a number is a multiple of $3$ or $5$ and then to include it in our sum.

```python
# Method 1 - looping through all numbers and checking if they are multiples
Sum = 0

for i in range(3, 1000):

    if i % 3 == 0 or i % 5 == 0:
        Sum += i

print('The sum of all multiples of 3 or 5 below 1000 is', Sum)

```

--------------------------------------------
Of course we can be more clever and use arithmetic series.

$$S_n = \frac{n}{2} (2 a_1 + (n-1)d),$$
where $n$ is the numnber of terms, $a_1$ is the first entry of the corresponding arithmetic sequence, and $d$ is the difference.

For our purposes $a_1$ and $d$ are the same so lets define

$$S_n(d) = \frac{n}{2}* (n+1)d$$

```python
#Method 2 - using arithmetic series

def Sum_multiples(d, N):
    """ Returns the sum of all multiples of d below n """
    
    # number of multiples of d below n
    n = (N-1) // d  
    
    # use formula for sum of arithmetic series
    S = d * n * (n + 1) / 2  
    
    return S

Sum = Sum_multiples(3, 1000) + Sum_multiples(5, 1000) - Sum_multiples(15, 1000)

print('The sum of all multiples of 3 or 5 below 1000 is', int(Sum))
```

### Exercise 4.6
Use Euclid's algorithm to count how many divisions are required to compute the greatest common divisor of $a$ and $b$ where
$$a = 45397369416530795319729696969741061923382 $$
and
$$b = 280571172992510140037611932413038677189525. $$


---------------------------
Idea of Euclid's algorithm follows from following theorem

Suppose $m \ge n > 0$ are natural numbers, and that $m= qn + r$ with $0 \ge r < n$. Then 
$$ \text{hcf}(m,n) = \text{hcf}(n,r).$$

```python
a = 45397369416530795319729696969741061923382 
b = 280571172992510140037611932413038677189525
count = 0
# Loop until the remainder on division is zero
while b != 0:
    (a, b) = (b, a % b)
    count += 1


print('Number of divisions required to compute gcd is', count)

print('The hcf of a and b is', a)
```

### Exercise 4.8
Adapt the bisection code example to compute an approximation to ten decimal digits of $\pi$.

```python

import numpy as np

# Set up: choose function and starting values
def f(x):
    return (x - np.pi)

(a, b) = (0, 4)    # starting guess for the root 

assert f(a) * f(b) < 0  # check that f(a) and f(b) have opposite sign

# Calculate first midpoint
c = (a + b) / 2

iteration = 0  # iteration counter
print(f"{iteration}: c = {c}, f(c) = {f(c)}")
while True:
    if f(c) * f(a) < 0:
        # f(c) has the opposite sign to f(a), so c replaces b
        (a, b) = (a, c)
    else:
        # f(c) must have the opposite sign to f(b), so c replaces a
        (a, b) = (c, b)

    # Calculate midpoint for the next round of the loop
    c = (a + b) / 2

    # Increment counter
    iteration += 1
    #print(f"{iteration}: c = {c}, f(c) = {f(c)}")

    if abs(f(c)) < 1e-10:
        # Close enough, terminate with success
        break

    if iteration > 100:
        # Too many iterations, terminate with failure
        break
    
print('Approximation to pi is', f"{c:.10f}")
```
