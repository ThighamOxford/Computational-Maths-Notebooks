### Exercise 10.1

Consider a curve parametried by $t \in [0,1]$, given by
$$ (x,y,z) = (0, t, t^2).$$
Calculate the arclength $L$ of this curve with 
$$ L = \int_0^1 \sqrt{(\mathrm{d}x/ \mathrm{d}t)^2 + (\mathrm{d}y/ \mathrm{d}t)^2 + (\mathrm{d}z /\mathrm{d}t)^2}\mathrm{d}t.$$

Aside: why is this the correct integral?

```python
# Define parameter t
t = sp.Symbol("t", nonnegative = True)

# Define x, y, z in terms of t
(x,y,z) = (0*t, t, t**2)
(xdot, ydot, zdot) = ( sp.diff(x,t), sp.diff(y,t), sp.diff(z,t) )

# Integrate
L = sp.integrate(sp.sqrt(xdot**2 + ydot**2 + zdot**2), (t, 0, 1))
print("The arclength L is", L)
```

### Exercise 10.3
A falling object $o$ encounters a moving platform $p$ accelerating upwards. Their heights for a given time $t \ge 0$ are
$$ h_o(t) = h_0 - v_0 t - \frac{1}{2} g t^2,$$
$$ h_p(t) = v_p t + \frac{1}{2} q t^2,$$

where $v_0, v_p \ge 0$ are the initial velocities, and $g, q > 0 $ are the (constant) accelerations. Find the initial velocity $v_0$ such that when the object and platform collide, they are moving at the same speed.

```python
# Define symbols
(t, h0, v0, vp, g, q) = sp.symbols("t, h0, v0, vp, g, q", nonnegative=True)

# First look for time of collision

h_obj = h0 - v0 * t - sp.Rational(1,2) * g * t**2
h_plat = vp * t + sp.Rational(1,2) * q * t**2
t_col = sp.solve( h_obj - h_plat, t) #this gives one solution
#print(t_col)


# Solve for when derivatives are equal in magnitude at this time
v_obj = sp.diff(h_obj,t)
v_plat = sp.diff(h_plat,t)
eq = sp.Eq(v_obj.subs(t, t_col[0]) + v_plat.subs(t, t_col[0]), 0)

init_speed = sp.solve( eq, v0)
print("v0 is given by \n", init_speed)
```

```python
"""Plugging in some values"""
speed = [sol.subs({h0:100, vp:50, g:9.8, q:1}) for sol in init_speed]
print(speed)
```

Solving differential equations

$$ x f''(x) + f'(x) = x^3 $$
subject to
$$ f(1) = 0, \quad f'(2) = 1.$$

-----------------------------------------

Things to do
- represent x and f symbolically. Also derivatives of f using f.diff(x,n)
- symbolically form the equation to be solved.
- solve the equation (with constants of integration)
- Sub in the initial conditions we have.


```python
# Symbols
x = sp.Symbol("x")
f = sp.Function("f")(x) 

# Equation
eq = sp.Eq(x*f.diff(x, 2) + f.diff(x), x**3) # comma separates LHS and RHS

# Pass to solver
sol = sp.dsolve(eq, f) # this returns a sympy equation
print("The sympy equation looks like", sol)

# Pass to solver with initial conditions
ics = {f.subs({x: 1}): 0, f.diff(x).subs({x: 2}): 1} # Two levels of dictionary
print("The solution is") 
sp.dsolve(eq, f, ics=ics) #ics is an optional argument
```
